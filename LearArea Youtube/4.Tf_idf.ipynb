```python
import pandas as pd
```

```python
df=pd.read_csv("../Dataset/IMDB Dataset.csv")
```

```python
df.head()
```

```python
def lowering(text):
    return text.lower()


df["review"]=df["review"].apply(lowering)
```

```python
from bs4 import BeautifulSoup

def remove_html_tags(text):
    return BeautifulSoup(text, "html.parser").get_text()

df["review"]=df["review"].apply(remove_html_tags)
```

```python
import string,time



def remove_punctuation(value):
    return value.translate(str.maketrans('','',string.punctuation))

df["review"]=df["review"].apply(remove_punctuation)
```

```python
from nltk.corpus import stopwords

stop=stopwords.words("english")
```

```python
def stopword_removal(text):
    ll=[]
    for word in text.split():
        if word not in stop:
            ll.append(word)
    return " ".join(ll) 
```

```python
df["review"]=df["review"].apply(stopword_removal)
df.head()
```

```python
from nltk import WordNetLemmatizer

lemmatizer=WordNetLemmatizer()

def lemma(text):
    ll=[]
    for word in text.split():
        ll.append(lemmatizer.lemmatize(word))
    return " ".join(ll) 
```

```python
df["review"]=df["review"].apply(lemma)
```

```python
def convert_label(x):
    if(x=="positive"):
        return 1
    else:
        return 0
    
df["sentiment"]=df["sentiment"].apply(convert_label)
```

```python
X=df["review"]
y=df["sentiment"]
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=.25)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer=TfidfVectorizer(max_features=10000)
```

```python
transformed_train=vectorizer.fit_transform(x_train)
```

```python
print(vectorizer.vocabulary_)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
from sklearn.naive_bayes import MultinomialNB

model=MultinomialNB()
```

```python
model.fit(transformed_train,y_train)
```

```python
transformed_test=vectorizer.transform(x_test)
```

```python
pred=model.predict(transformed_test)
```

```python
from sklearn.metrics import classification_report

print(classification_report(pred,y_test))
```
