```python
import pandas as pd
```

```python
df=pd.read_csv("../Dataset/IMDB Dataset.csv")
df.head()
```

```python
from nltk import word_tokenize

tokens=word_tokenize(df["review"][0])
```

```python
from nltk import PorterStemmer,WordNetLemmatizer

stemmer=PorterStemmer()
lemmatizer=WordNetLemmatizer()
```

```python
stemmed=[]
for i in tokens:
    stemmed.append(stemmer.stem(i))
```

```python
stemmed
```

```python
lemmad=[]
for i in tokens:
    lemmad.append(lemmatizer.lemmatize(i))
```

```python
lemmad
```

```python
for i in range(len(tokens)):
    print(tokens[i]+"  ->  "+stemmed[i]+"  ->  "+lemmad[i])
```
