```python
import pandas as pd
import numpy as np

df=pd.read_csv("../Dataset/Twitter_Data.csv")
df.head()
```

```python
from gensim.utils import simple_preprocess


df["clean_text"]=df["clean_text"].astype("str").apply(simple_preprocess)
df.head()
```

```python
def join_lines(list_value):
    text=" ".join(list_value)
    return text

df["clean_text"]=df["clean_text"].apply(join_lines)
```

```python
Dataset="__label__"+df["category"].astype("str")+" "+df["clean_text"]
```

```python
Dataset[:10]
```

```python
from sklearn.model_selection import train_test_split

train,test=train_test_split(Dataset,random_state=42,test_size=.25)
```

```python
display(train.shape)
display(test.shape)
```

## Fasttext requires text/csv format for train and test

```python
train_file=train.to_csv("../Resource/train.text",index=False,header=False)
test_file=train.to_csv("../Resource/test.text",index=False,header=False)
```

```python
trained_model=fasttext.train_supervised(input="../Resource/train.text",epoch=25, lr=1.0, wordNgrams=2)
```

```python
test_result=trained_model.test("../Resource/test.text")
```
